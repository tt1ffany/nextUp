```python
raw_documents = TextLoader("books_tagged_description.txt", encoding="utf-8").load()

# Set chunk size to 1 to prioritize splitting by separator (new line) instead of chunk size
text_splitter = CharacterTextSplitter(chunk_size=1, chunk_overlap=0, separator="\n")
documents = text_splitter.split_documents(raw_documents)
```

```python
# Use HuggingFace embeddings instead of Gemini API to avoid rate limits
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embedding_function = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

db_books_hf = Chroma.from_documents(
    documents,
    embedding=embedding_function,
    persist_directory="./chroma_db_hf"
)
```

```python
query = "Dystopian corporate surveillance and the loss of individual privacy"
docs = db_books_hf.similarity_search(query, k=10)
docs
```

```python
# Filter books DataFrame to find the book(s) corresponding to the first document in the search results
books[books["isbn13"] == int(docs[0].page_content.split()[0].strip())]
```

```python
# Function to retrieve semantic recommendations based on a query
def retrieve_semantic_recommendations(
        query: str,
        top_k: int = 10,   
) -> pd.DataFrame:
    recs = db_books_hf.similarity_search(query, k=50)
    books_list = []

    for i in range(0, len(recs)):
        books_list += [int(recs[i].page_content.strip('"').split()[0])]

    return books[books["isbn13"].isin(books_list)]
```

```python
retrieve_semantic_recommendations("Dystopian corporate surveillance and the loss of individual privacy")
```
