```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
books = pd.read_csv(f"{path}/books.csv")
```

```python
books
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)

plt.xlabel("Columns")
plt.ylabel("Missing Values")
plt.show()
```

```python
import numpy as np

books["missing_description"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"] = 2026 - books["published_year"]
```

```python
columns_of_interest = ["num_pages", "age_of_book", "missing_description", "average_rating"]

correlation_matrix = books[columns_of_interest].corr(method="spearman")

sns.set_theme(style="white")
plt.figure(figsize=(8, 6))
heatmap = sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", cbar_kws={"label": "Spearman Correlation Coefficient"})

heatmap.set_title("Spearman Correlation Heatmap")
plt.show()
```

```python
# Keep books that have no missing values in the columns of interest
book_missing = books[~(books["description"].isna()) &
                      ~(books["num_pages"].isna()) &
                      ~(books["average_rating"].isna()) &
                      ~(books["published_year"].isna())
]
```

```python
book_missing
```

```python
# Display the count of books in each category
book_missing["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
# Create new variable for the number of words in the descriptions (descriptions with very few words may not be very informative/meaningful)
book_missing["words_in_description"] = book_missing["description"].str.split().str.len()
```

```python
# Display with new variable "words_in_description"
book_missing
```

```python
# Smallest descriptions grouping (1 to 4 words)
book_missing.loc[book_missing["words_in_description"].between(1, 4), ["description"]]
```

```python
# 5 to 14 words descriptions grouping
book_missing.loc[book_missing["words_in_description"].between(5, 14), ["description"]]
```

```python
# 15 to 24 words descriptions grouping
book_missing.loc[book_missing["words_in_description"].between(15, 24), ["description"]]
```

```python
# 25 to 34 words descriptions grouping (will start to see descriptions that are more informative/meaningful)
book_missing.loc[book_missing["words_in_description"].between(25, 34), ["description"]]
```

```python
# Keep books with descriptions that have 25 or more words
book_missing_25_words = book_missing[book_missing["words_in_description"] >= 25]
```

```python
book_missing_25_words
```

```python
# If subtitle is missing, just use the title, otherwise combine the title and subtitle into a new variable called "title_subtitle"

book_missing_25_words["title_subtitle"] = (
    np.where(book_missing_25_words["subtitle"].isna(), book_missing_25_words["title"],
             book_missing_25_words["title"] + ": " + book_missing_25_words["subtitle"])
)
```

```python
book_missing_25_words
```

```python
# Create a new variable called "tagged_description" that combines the ISBN13 and description into one string (useful for tagging and searching purposes)
book_missing_25_words["tagged_description"] = (
    book_missing_25_words["isbn13"].astype(str).str.cat(book_missing_25_words["description"], sep=" ")
)
```

```python
book_missing_25_words
```

```python
# Drop columns that are no longer needed
columns_to_drop = ["subtitle", "missing_description", "words_in_description", "age_of_book"]
(
    book_missing_25_words.drop(columns_to_drop, axis=1).to_csv("books_cleaned.csv", index=False)
)
```
